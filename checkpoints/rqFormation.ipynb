# Project Scope

**Dataset:** Airbnb Prices in European Cities 

**Key EDA Findings (Amsterdam subset):**
- Price (`realSum`) is highly skewed with significant outliers
- Strong relationships observed:
  - Distance negatively correlates with attraction index
  - Room type significantly impacts price
  - Capacity affects price in a nonlinear way
- Spatial patterns suggest clustering based on location

**Planned Techniques:**
- **Course Techniques:** Clustering, Anomaly Detection  
- **External Techniques:** Spatial modeling, advanced regression (e.g., Gradient Boosting)


# Research Question Definition

## RQ1: How do listings cluster based on price, location, and property features?
- **Task Type:** Clustering  
- **Algorithms:** K-Means, DBSCAN  
- **Metrics:** Silhouette Score, cluster interpretability  

## RQ2: What listings are anomalous in terms of price relative to their features?
- **Task Type:** Anomaly Detection  
- **Algorithms:** Isolation Forest, Z-score (baseline)  
- **Metrics:** Number of anomalies detected, interpretability

## RQ3: Does spatial location (latitude/longitude + attraction index) improve price prediction compared to non-spatial features?
- **Task Type:** Regression / Spatial Modeling  
- **Algorithms:**  
  - Baseline: Linear Regression  
  - External: Gradient Boosting (e.g., XGBoost / LightGBM)  
- **Metrics:** RMSE, R²

# Motivation and Feasibility

**Motivation:**
- EDA reveals strong clustering behavior and spatial patterns
- Price is influenced by multiple interacting features (location, capacity, room type)

**Non-triviality:**
- Basic statistical analysis does not capture:
  - Nonlinear relationships  
  - Spatial dependencies  
- Advanced methods are needed to uncover deeper structure

**Feasibility:**
- Dataset size is manageable using pandas and scikit-learn
- Required libraries are readily available:
  - `sklearn` (KMeans, IsolationForest, Linear Regression)
  - `xgboost` / `lightgbm` for advanced regression

**Risks:**
- Price skewness may affect modeling → may require log transformation  
- DBSCAN is sensitive to parameter tuning  
- Spatial clustering may introduce noise  

# Methodological Plan

**Clustering:**
- Normalize features
- Apply K-Means with varying cluster sizes (k)
- Apply DBSCAN to capture spatial clusters

**Anomaly Detection:**
- Run Isolation Forest
- Compare results with Z-score baseline

**Regression (External):**
- Train baseline Linear Regression model
- Train Gradient Boosting model
- Compare performance with and without spatial features
